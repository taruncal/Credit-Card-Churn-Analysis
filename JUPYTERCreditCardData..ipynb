```python
import pandas as pd

# 1. Age Groups
bins = [20, 30, 45, 60, 100]
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['Age_Group'] = pd.cut(df['Customer Age'], bins=bins, labels=labels)

# 2. Tenure in Years
df['Tenure_Years'] = df['Months on Book'] / 12

# 3. Average Transaction Value
df['Avg_Transaction_Value'] = df['Total Transaction Amount'] / df['Total Transaction Count']

# 4. Credit Utilization Tier
df['Utilization_Tier'] = pd.cut(
    df['Average Utilization Ratio'], 
    bins=[-0.1, 0.3, 0.7, 1.0], 
    labels=['Low', 'Moderate', 'High']
)
```

```python
print(df[['Customer Age', 'Age_Group', 'Months on Book', 'Tenure_Years']].head(9))
```

```python
df = df.rename(columns={
    'Age_Group': 'Age Group',
    'Tenure_Years': 'Tenure Years',
    'Avg_Transaction_Value': 'Avg Transaction Value',
    'Utilization_Tier': 'Utilization Tier'
})
```

```python
df['Tenure Years'] = df['Tenure Years'].round(2)
df['Avg Transaction Value'] = df['Avg Transaction Value'].round(2)
```
